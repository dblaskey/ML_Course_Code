**Goal:** Open and analyze a dataset.

The Meuse dataset contains point observations of topsoil heavy-metal concentrations from a floodplain along the Meuse River near Stein, Netherlands. It includes spatial coordinates, heavy-metal concentrations such as zinc, cadmium, copper, and lead, plus covariates such as distance to river, elevation-related variables, soil type, and flooding frequency. Contaminated sediment is transported by the river and tends to be deposited closer to the river bank and in lower-elevation floodplain areas.

## Step 2: Create a clean project structure

A consistent folder structure makes projects easier to understand, rerun, share, and debug.

We will create four main directories:

- `data/raw/` for original input data
- `notebooks/` for exploratory notebooks
- `outputs/` for generated files, figures, and results
- `scripts/` for reusable Python scripts

````python
from pathlib import Path

folders = ["data/raw", "notebooks", "outputs", "scripts"]

for folder in folders:
    Path(folder).mkdir(parents=True, exist_ok=True)

!find . -maxdepth 3 -type d | sort
````

## Step 5: Create a `.gitignore` file

Not every file belongs in Git.

For example, we usually do **not** commit:

- Virtual environments
- Python cache files
- Notebook checkpoint files
- Generated outputs
- System files like `.DS_Store`

````python
gitignore_text = ".venv/
__pycache__/
.ipynb_checkpoints/
outputs/
.DS_Store
"
Path(".gitignore").write_text(gitignore_text)
!cat .gitignore
````

````python
!git status
````
